```python
import pandas as pd
```

```python
df = pd.read_csv("nba_games.csv", index_col=0) # starts at column 0
```

```python
df = df.sort_values("date") # sorts data frame by a single column in order of date
```

```python
df = df.reset_index(drop=True) # resests index back to 0 and drops old index
```

```python
del df["mp.1"]
del df["mp_opp.1"]
del df["index_opp"]  # deletes certain columns
```

```python
def add_target(team): # 
    team["target"] = team["won"].shift(-1)
    return team

df = df.groupby("team", group_keys=False).apply(add_target) # creates data frame for each team
```

```python
df["team"] # teams list
df[df["team"] == "WAS"] # marksdown to specific team
```

```python
df["target"][pd.isnull(df["target"])] = 2 # takes target, uses isnull func and replaces them with 2
```

```python
df["target"] = df["target"].astype(int, errors="ignore") # converts boolean into ints
```

```python
df["won"].value_counts()
```

```python
df["target"].value_counts()
```

```python
nulls = pd.isnull(df).sum()
nulls = nulls[nulls > 0]
valid_columns = df.columns[~df.columns.isin(nulls.index)]
valid_columns
df = df[valid_columns].copy()
```

```python
from sklearn.model_selection import TimeSeriesSplit # will help split data
from sklearn.feature_selection import SequentialFeatureSelector
from sklearn.linear_model import RidgeClassifier # machine learning model

rr = RidgeClassifier(alpha=1)
split = TimeSeriesSplit(n_splits=3)

sfs = SequentialFeatureSelector(rr, n_features_to_select=30, direction="forward", cv=split)
```

```python
removed_columns = ["season", "date", "won", "target", "team", "team_opp"]
```

```python
selected_columns = df.columns[~df.columns.isin(removed_columns)] # ~deletes columns under certain condition
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
df[selected_columns] = scaler.fit_transform(df[selected_columns])
```

```python
sfs.fit(df[selected_columns], df["target"])
```

```python
predictors = list(selected_columns[sfs.get_support()]) # sfs.get_support is a boolean True or false that indexes values that are going to be used to train the model
```
